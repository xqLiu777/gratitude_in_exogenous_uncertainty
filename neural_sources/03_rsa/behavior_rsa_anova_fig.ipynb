```python
import seaborn as sns
import numpy as np
import os
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# 通过工作表名称读取
df = pd.read_excel(file_path, sheet_name='inten_HSD')

# 绘图
sns.set_theme(style="white")

# Set up the matplotlib figure
f, ax = plt.subplots(figsize=(130, 100))

# Generate a custom diverging colormap
cmap = sns.color_palette("OrRd", 100)

# Draw the heatmap with the mask and correct aspect ratio
sns_plot = sns.heatmap(df, cmap=cmap, square=True, linewidths=.5, cbar = True, mask=mask, 
                       annot=True, fmt='.3f', annot_kws={'size': 400, 'color': 'black'},
                       cbar_kws={'shrink': 0.9, 'label': 'Color Scale', 'ticks': np.arange(0, 1.1, 0.2), 'pad': 0.1})
# , vmax = 1.1, vmin = -0.1

cax = plt.gcf().axes[-1]
cax.tick_params(labelsize = 400, pad = 50)

# 设置 x 轴和 y 轴的刻度标签
plt.xticks([0.5, 1.5, 2.5, 3.5], labels = ['H1', 'H2', 'H3', 'H4'], fontsize = 400)
plt.yticks([0.5, 1.5, 2.5, 3.5], labels = ['H1', 'H2', 'H3', 'H4'], fontsize = 400)

# 增大刻度标签的距离
plt.tick_params(axis='x', pad = 100)  # x轴刻度标签的距离
plt.tick_params(axis='y', pad = 100)  # y轴刻度标签的距离

ax.set_yticklabels(ax.get_yticklabels(), rotation=0)
# ax.set_xticklabels(ax.get_xticklabels(), rotation=20)

# # 调整页边距，参数分别为 left, bottom, right, top
# plt.subplots_adjust(left=0.1, right=0.9, top=0.9, bottom=0.1)

# plt.legend(fontsize = 18)  # 将字体大小设置为12

# plt.show()

plt.tight_layout()

plt.savefig(fig_dir + 'intention_HSD_250104.svg', format='svg', dpi = 200)

# plt.savefig(fig_dir + 'intention_rdm_241114.png', dpi = 200)
plt.show()
```
