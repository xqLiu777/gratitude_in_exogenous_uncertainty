```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from itertools import cycle
```

```python
# 读取CSV文件
raw_dir = 'E:\\lxq\\jupyter\\CMF2024\\fmri\\03_rsa\\rsa_rho\\'
themes = ["Gratitude","Allocation", "Intention", "Cost"]
```

```python

rois = [52, 3, 148, 98, 136, 27, 194, 11]

for roi in rois:
    r_df = pd.DataFrame()
    for theme in themes: 
        data = np.loadtxt(raw_dir + 'WSRSA_' + theme + '_200_r_value.txt')
        tmp_r_df = pd.DataFrame(data[roi - 1], columns=['r_value'])

        # 添加计数列
        tmp_r_df['subID'] = np.arange(1, 47)  # 计数从1到50

        # 添加标记种类列（假设有两种类别A和B，交替标记）
        tmp_r_df['theme'] = theme

        # 添加新行
        r_df = pd.concat([r_df, tmp_r_df], ignore_index=True)
    
    r_df.to_excel(raw_dir + "ROI_" + str(roi) + '_rho_value_241107.xlsx', index=False)
    
```
